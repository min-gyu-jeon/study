```python
def select_random(min, max):
    import random
    n = random.randint(min, max)
    return n                          #n을 함수를 호출한 곳으로 반환

def show_number():
    n = select_random(50, 150)        #함수를 호출하고 함수가 반환한 값을 n에 저장
    print("select number : ", n)

show_number()
    
```

```python
def get_stats(numbers):
    total = sum(numbers)
    average = total / len(numbers)
    maximum = max(numbers)
    minimum = min(numbers)
    return total, average, maximum, minimum
        
```

```python
numbers = [1, 2, 3, 4, 5]
t, a, mx, mn = get_stats(numbers)
print(t, a, mx, mn)

stats = get_stats(numbers)
print(stats)
print(stats[0], stats[1], stats[2], stats[3])
```

```python
#가변인자 배열 (전달인자의 갯수와 관계없이 튜플 형태로 수신)

def total(*args):
    result = 0
    for num in args:
        result += num
    return result

print(total(1, 2, 3))           # 6
print(total(10, 20, 30, 40))    # 100
print(total(5))                 # 5
```

```python
# 일반 함수
def add(a, b):
    return a + b

add2 = add                          #함수는 1급 객체이므로 다른 변수에 저장 가능

# 람다 함수 (동일한 기능)

add_lambda = lambda a, b: a + b

print(add(3, 5))         # 8
print(add2(3, 5))
print(add_lambda(3, 5))  # 8
```

```python
# ── 함수 정의 ──────────────────────────────

def get_grade(score):
    """점수를 받아 학점 문자를 반환합니다."""
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

def get_stats(scores):
    """점수 리스트를 받아 (평균, 최고, 최저)를 반환합니다."""
    average = sum(scores) / len(scores)
    return average, max(scores), min(scores)

def print_report(students):
    """학생 목록을 받아 성적표를 출력합니다."""
    print("=" * 45)
    print(f"  {'이름':<8} {'점수':>5} {'학점':>5} {'결과':>6}")
    print("-" * 45)

    scores = []
    for name, score in students:
        grade = get_grade(score)
        result = "합격" if score >= 60 else "불합격"
        print(f"  {name:<8} {score:>5} {grade:>5} {result:>6}")
        scores.append(score)

    avg, high, low = get_stats(scores)

    print("=" * 45)
    print(f"  수강 인원: {len(students)}명")
    print(f"  평균: {avg:.1f}점  |  최고: {high}점  |  최저: {low}점")

    # 상위 학생 (람다 활용 정렬)
    top_students = sorted(students, key=lambda x: x[1], reverse=True)
    print(f"\n  🏆 1등: {top_students[0][0]} ({top_students[0][1]}점)")
    print("=" * 45)


# ── 메인 코드 ──────────────────────────────

students = [
    ("홍길동", 92),
    ("김철수", 75),
    ("이영희", 88),
    ("박민준", 55),
    ("최지은", 67),
    ("정수현", 98),
]

print_report(students)
```
