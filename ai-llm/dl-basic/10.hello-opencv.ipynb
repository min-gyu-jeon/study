```python
import cv2
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
```

```python
cv_img = cv2.imread('images/building.jpg')
pil_img=Image.open('images/building.jpg')
```

```python
# plt.imshow(cv_img)
plt.imshow(pil_img)
plt.axis('off')
plt.show()
```

```python
cv_img = cv2.imread('images/building.jpg')
cv2.rectangle(cv_img, (350, 150), (550, 450), (0, 255, 0), 2)
cv2.putText(cv_img, "Hello, OpenCV", (350, 140), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
plt.imshow(cv_img)
plt.axis('off')
plt.show()
```

```python
# cv_img = cv2.imread('images/building.jpg')

print( cv_img.shape )
height, width, _ = cv_img.shape
center = ( width // 2, height // 2 )
# center = ( 100, 100 )
rotate = cv2.getRotationMatrix2D(center, -45, 1) #45도 회전을 위한 매트릭스 계산
print(rotate)
rotated = cv2.warpAffine(cv_img, rotate, (width, height))   #변환 처리 

plt.imshow(rotated)
plt.axis('off')
plt.show()
```

```python
# 이미지 flip (좌우, 상하 반전)

flipped = cv2.flip(cv_img, -1) # 1(좌우), 0(상하), -1(상하/좌우)
plt.imshow(flipped)
plt.axis('off')
plt.show()
```

```python
# 이미지 크기 변경
print( cv_img.shape )
height, width, _ = cv_img.shape
ratio = 400 / width # 확대 or 축소 비율
t_size = ( 400, int(height * ratio ) )
# t_size = ( 400, 150 )

resized = cv2.resize(cv_img, t_size, cv2.INTER_AREA)
print( resized.shape )

plt.imshow(resized)
plt.axis('off')
plt.show()
```

```python
(height, width) = cv_img.shape[:2]
center = (width // 2, height // 2)

mask = np.zeros(cv_img.shape[:2], np.uint8) # 검정색 사각형 이미지 (배경)
cv2.circle(mask, center, 200, 255, -1) # 흰색 원 그리기

plt.imshow(mask, cmap='gray')
plt.axis('off')
plt.show()
```

```python
masked = cv2.bitwise_and(cv_img, cv_img, mask=mask) # mask의 흰색 영역만 표시

plt.imshow(masked)
plt.axis('off')
plt.show()
```
